```python
df['year']=pd.to_datetime(df['release_date'],errors='coerce').apply(lambda x: str(x).split('-')[0] if x != np.nan else np.nan)
```

```python
qualified = df[(df['vote_count'] >= m) & (df['vote_count'].notnull()) & (df['vote_average'].notnull())][['title', 'year', 'vote_count', 'vote_average', 'popularity', 'genres']]
qualified['vote_count'] = qualified['vote_count'].astype('int')
qualified['vote_average'] = qualified['vote_average'].astype('int')
qualified.shape
```

```python
def weighted_rating(x):
    v = x['vote_count']
    R = x['vote_average']
    return (v/(v+m) * R) + (m/(m+v) * C)
```

```python
qualified['score'] = qualified.apply(weighted_rating, axis=1)
```

```python
qualified = qualified.sort_values('score', ascending=False).head()
```

```python
qualified.head()
```

```python
s = df.apply(lambda x: pd.Series(x['genres']),axis=1).stack().reset_index(level=1, drop=True)
s.name = 'genre'
gen_md = df.drop('genres', axis=1).join(s)
```

```python
gen_md.head()
```

```python
def build_chart(genre, percentile=0.85):
    d = gen_md[gen_md['genre'] == genre]
    vote_counts = d[d['vote_count'].notnull()]['vote_count'].astype('int')
    vote_averages = d[d['vote_average'].notnull()]['vote_average'].astype('int')
    C = vote_averages.mean()
    m = vote_counts.quantile(percentile)
    
    qualified = d[(d['vote_count'] >= m) & (d['vote_count'].notnull()) & (d['vote_average'].notnull())][['title', 'year', 'vote_count', 'vote_average', 'popularity']]
    qualified['vote_count'] = qualified['vote_count'].astype('int')
    qualified['vote_average'] = qualified['vote_average'].astype('int')
    
    qualified['wr'] = qualified.apply(lambda x: (x['vote_count']/(x['vote_count']+m) * x['vote_average']) + (m/(m+x['vote_count']) * C), axis=1)
    qualified = qualified.sort_values('wr', ascending=False).head(250)
    
    return qualified
```

<h1>Content Based Recommender</h1>

```python
links_small = pd.read_csv('dataset/links_small.csv')
links_small = links_small[links_small['tmdbId'].notnull()]['tmdbId'].astype('int')
```

```python
df = df.drop([19730, 29503, 35587])
```

```python
df['id'] = df['id'].astype('int')
```

```python
smd = df[df['id'].isin(links_small)]
smd.shape
```

```python
# Fill NaN values in 'tagline' and 'overview' columns with an empty string
smd.loc[:, 'tagline'] = smd['tagline'].fillna('')

# Create the 'description' column by combining 'overview' and 'tagline'
smd.loc[:, 'description'] = smd['overview'] + smd['tagline']

# Fill NaN values in 'description' with an empty string
smd.loc[:, 'description'] = smd['description'].fillna('')
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer, CountVectorizer
tf = TfidfVectorizer(analyzer='word',ngram_range=(1, 2),min_df=1, stop_words='english')
tfidf_matrix = tf.fit_transform(smd['description'])
```

```python
tfidf_matrix.shape
```

```python
from sklearn.metrics.pairwise import linear_kernel, cosine_similarity
cosine_sim = linear_kernel(tfidf_matrix, tfidf_matrix)
```

```python
cosine_sim[0]
```

```python
smd = smd.reset_index()
titles = smd['title']
indices = pd.Series(smd.index, index=smd['title'])
```

```python
def get_recommendations(title):
    idx = indices[title]
    sim_scores = list(enumerate(cosine_sim[idx]))
    sim_scores = sorted(sim_scores, key=lambda x: x[1], reverse=True)
    sim_scores = sim_scores[1:31]
    movie_indices = [i[0] for i in sim_scores]
    return titles.iloc[movie_indices]
```

<h1>Metadata Based Recommender
</h1>

```python
credits = pd.read_csv('dataset/credits.csv')
keywords = pd.read_csv('dataset/keywords.csv')
```

```python
keywords['id'] = keywords['id'].astype('int')
credits['id'] = credits['id'].astype('int')
df['id'] = df['id'].astype('int')
```

```python
df.shape
```

```python
df = df.merge(credits, on='id')
df = df.merge(keywords, on='id')
```

```python
smd = df[df['id'].isin(links_small)]
smd.shape
```

```python
from ast import literal_eval

# Check if the column contains strings and apply literal_eval only to string entries
def safe_literal_eval(value):
    try:
        if isinstance(value, str):
            return literal_eval(value)
        return value  # Return value as is if it's not a string
    except (ValueError, SyntaxError):
        return value  # Return value unchanged if it's not a valid literal

# Apply the function to the columns using .loc[] to avoid SettingWithCopyWarning
smd.loc[:, 'cast'] = smd['cast'].apply(safe_literal_eval)
smd.loc[:, 'crew'] = smd['crew'].apply(safe_literal_eval)
smd.loc[:, 'keywords'] = smd['keywords'].apply(safe_literal_eval)
```

```python
def get_director(x):
    for i in x:
        if i['job'] == 'Director':
            return i['name']
    return np.nan
```

```python
smd.loc[:, 'director'] = smd['crew'].apply(get_director)
```

```python

# Ensure 'cast' is evaluated into a list of dictionaries
smd['cast'] = smd['cast'].apply(lambda x: literal_eval(x) if isinstance(x, str) else x)

# Extract actor names and limit to top 3 actors
smd['cast'] = smd['cast'].apply(lambda x: [i['name'] for i in x if isinstance(i, dict) and 'name' in i] 
                                if isinstance(x, list) else [])
smd['cast'] = smd['cast'].apply(lambda x: x[:3] if len(x) >= 3 else x)
```

```python
smd['keywords'] = smd['keywords'].apply(lambda x: [i['name'] for i in x] if isinstance(x, list) else [])
```

```python
smd['cast'] = smd['cast'].apply(lambda x: [str.lower(i.replace(" ", "")) for i in x])
```

```python
smd['director'] = smd['director'].astype('str').apply(lambda x: str.lower(x.replace(" ", "")))
smd['director'] = smd['director'].apply(lambda x: [x,x, x])
```

```python
s = smd.apply(lambda x: pd.Series(x['keywords']),axis=1).stack().reset_index(level=1, drop=True)
s.name = 'keyword'
```

```python
s = s.value_counts()
s[:5]
```

```python
s = s[s > 1]
```

```python
from nltk.stem.snowball import SnowballStemmer
stemmer = SnowballStemmer('english')
stemmer.stem('dogs')
```

```python
def filter_keywords(x):
    words = []
    for i in x:
        if i in s:
            words.append(i)
    return words
```

```python
smd['keywords'] = smd['keywords'].apply(filter_keywords)
smd['keywords'] = smd['keywords'].apply(lambda x: [stemmer.stem(i) for i in x])
smd['keywords'] = smd['keywords'].apply(lambda x: [str.lower(i.replace(" ", "")) for i in x])
```

```python
smd['soup'] = smd['keywords'] + smd['cast'] + smd['director'] + smd['genres']
smd['soup'] = smd['soup'].apply(lambda x: ' '.join(x))
```

```python
count = CountVectorizer(analyzer='word',ngram_range=(1, 2),min_df=1, stop_words='english')
count_matrix = count.fit_transform(smd['soup'])
```

```python
cosine_sim = cosine_similarity(count_matrix, count_matrix)
```

```python
smd = smd.reset_index()
titles = smd['title']
indices = pd.Series(smd.index, index=smd['title'])
```

```python
get_recommendations('The Dark Knight').head(10)
```

```python
def improved_recommendations(title):
    idx = indices[title]
    sim_scores = list(enumerate(cosine_sim[idx]))
    sim_scores = sorted(sim_scores, key=lambda x: x[1], reverse=True)
    sim_scores = sim_scores[1:26]
    movie_indices = [i[0] for i in sim_scores]
    
    movies = smd.iloc[movie_indices][['title', 'vote_count', 'vote_average', 'year']]
    vote_counts = movies[movies['vote_count'].notnull()]['vote_count'].astype('int')
    vote_averages = movies[movies['vote_average'].notnull()]['vote_average'].astype('int')
    C = vote_averages.mean()
    m = vote_counts.quantile(0.60)
    qualified = movies[(movies['vote_count'] >= m) & (movies['vote_count'].notnull()) & (movies['vote_average'].notnull())]
    qualified['vote_count'] = qualified['vote_count'].astype('int')
    qualified['vote_average'] = qualified['vote_average'].astype('int')
    qualified['wr'] = qualified.apply(weighted_rating, axis=1)
    qualified = qualified.sort_values('wr', ascending=False).head(10)
    return qualified
```

```python
improved_recommendations('The Dark Knight')
```
